# General Statistics — Dataset Overview

This notebook provides a high-level summary of the dataset collected for the paper:

> **The Bitter Pill: Tracking and Remarketing on EU Pharmacy Websites**

It reports request counts, third-party and tracker prevalence, breakdowns by country, consent mode, and experiment mode, and top entities observed across 50 pharmacy websites in 5 EU countries.

**Data:** Expects HAR files under the path set in `common.BASE_HAR_DIR` (see `common.py`).
**Dependencies:** `pandas`, `seaborn`; project modules `utils`, `common`.

## Setup and data load

```python
import pandas as pd
import seaborn as sns
from utils import parse_har_files_to_dataframe
from common import BASE_HAR_DIR


sns.set_theme(style="whitegrid", context="paper")
pd.set_option('display.max_colwidth', 128)

df = parse_har_files_to_dataframe(BASE_HAR_DIR, test=False)
```

## Dataset summary

```python
print(f'Number of requests: {len(df)}')
print(f'Number of third-party requests: {len(df[df.third_party])}')
print(f'Number of tracking-related requests: {len(df[df.is_tracker])}')
print(f'Number of countries: {df.country.nunique()}')
print(f'Number of unique websites: {df.site_domain.nunique()}')
print(f'Number of unique request domains: {df.req_domain.nunique()}')
print(f'Number of unique CNAME domains: {df.cname.nunique()}')
```
